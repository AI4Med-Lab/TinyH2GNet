```python
import pandas as pd

# Path to your metrics file
csv_path = "/home/puneet/mk/code_model_training/modelResults/her2st_stnet_224/result_vit_2025-11-12_00-56-07/metrics_result.csv"

# Load the CSV
df = pd.read_csv(csv_path)

# Filter for validation phase only
val_df = df[df["phase"] == "val"]

# Metrics of interest
metrics = [
    "pearson_mean_cell_wise", "pearson_mean_genewise",
    "spearman_mean_cell_wise", "spearman_mean_genewise",
    "l1_error_mean", "l2_errors_mean"
]

# Compute per-fold mean and std
fold_stats = val_df.groupby("fold")[metrics].agg(["mean", "std"])

# Compute overall mean and std across all folds
overall_stats = {
    metric: {
        "overall_mean": val_df[metric].mean(),
        "overall_std": val_df[metric].std()
    } for metric in metrics
}

# Convert to DataFrame
overall_stats_df = pd.DataFrame(overall_stats).T

print("\n=== Per-Fold Mean & Std ===")
print(fold_stats)

print("\n=== Overall Mean & Std ===")
print(overall_stats_df)
```

```python
import numpy as np

teacher_params = count_params(teacher_model)

targets = {
    "Tiny-90": 90.0,
    "Tiny-95": 95.0,
    "Tiny-Max": 99.0,  # aspirational, not literal
}

phi_candidates = np.linspace(-2, -1000, 3000)

results = {k: {"phi": None, "reduction": 0.0, "params": None, "err": 1e9} for k in targets}

for phi in phi_candidates:
    student = EfficientNetTinyStudent(num_genes=845, phi=float(phi))
    p = count_params(student)
    reduction = (1 - p / teacher_params) * 100

    for name, target in targets.items():
        err = abs(reduction - target)
        if err < results[name]["err"]:
            results[name] = {
                "phi": float(phi),
                "reduction": reduction,
                "params": p,
                "err": err
            }

for name, res in results.items():
    print(f"{name}: phi={res['phi']:.2f}, params={res['params']:,}, reduction={res['reduction']:.2f}%")
```
